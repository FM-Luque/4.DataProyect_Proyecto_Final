#### 4️⃣ Unir  `df_2022, df_2023, df_2024, df_2025` del archivo `ventas.csv`

Como las 4 hojas tienen exactamente las mismas columnas y representan el mismo tipo de información, las unimos verticalmente con pd.concat().

Antes de unirlas, añadimos una columna `hoja_origen` para no perder la trazabilidad de qué hoja proviene cada registro (nos sera útil más adelante).


```python
# Añadimos una columna que identifique el año/hoja de origen antes de concatenar
df_2022["hoja_origen"] = "2022"
df_2023["hoja_origen"] = "2023"
df_2024["hoja_origen"] = "2024"
df_2025["hoja_origen"] = "2025"
```

```python
# Verificamos que se ha añadido la columna "hoja_origen"
#  Dimensiones del dataset
for año, df in [("2022", df_2022), ("2023", df_2023), ("2024", df_2024), ("2025", df_2025)]:
    print(f"Año {año}: tiene {df.shape[0]} Filas y {df.shape[1]} Columnas ")
    
```

```python
# Nombres de las columnas
for año, df in [("2022", df_2022), ("2023", df_2023), ("2024", df_2024), ("2025", df_2025)]:
    print(f"Columnas de {año}: {df.columns.tolist()}")
```

#### Concatenamos 

```python
df_ventas = pd.concat([df_2022, df_2023, df_2024,df_2025], ignore_index=True)
```

#### 5️⃣ Unir `df_ventas` con `df_productos`

Como hemos observado ambos comparten la columna `producto_id`, por lo que uniremos con `merge()` y con `how="left"` para conservar todas las dilas de `df_ventas`

```python
# Comprobamos si hay algún product_id en df_ventas que no exista en df_productos
set(df_ventas["product_id"]) - set(df_productos["product_id"])
```

```python
df_ventas_productos = df_ventas.merge(df_productos, on="product_id", how="left")
```

```python
print(len(df_ventas), len(df_ventas_productos))  # deberían ser iguales
df_ventas_productos.isnull().sum()  # si aparecen nulos en columnas de producto, algo no casó bien
```

#### 6️⃣ Unimos df_ventas_productos con df_clientes

Como hemos observado ambos comparten la columna `id_`, por lo que uniremos con `merge()` y con `how="left"` para conservar todas las dilas de `df_ventas_productos`

```python
df_clientes.info()
```

```python
df_ventas_productos.info()
```

```python
# Unir con df_clientes por la columna id_
df_final = df_ventas_productos.merge(df_clientes, on="id_", how="left")
```
